```python
# These columns describe house characteristics like access to main road and furnishing status.
categorical_cols = ["mainroad", "guestroom", "basement",
                    "hotwaterheating", "airconditioning",
                    "prefarea", "furnishingstatus"]
#use value_counts to understand category balance,
for col in categorical_cols:
    print(f"\nValue counts for {col}:")
    print(df[col].value_counts())
```

```python
numeric_cols = ["price", "area", "bedrooms", "bathrooms", "stories", "parking"]
df[numeric_cols].describe()
```

## Visualize

Create any visualizations for your EDA here. Make note in the form of code comments of what your thought process is for your visualizations.

```python
# 1) Price distribution
# Thought process: understand how house prices are distributed and whether the distribution is skewed.
plt.figure()
sns.histplot(df["price"], kde=True)
plt.title("Distribution of House Prices")
plt.xlabel("Price")
plt.ylabel("Count")
plt.show()
```

```python
# 2) Area distribution
# Thought process: see whether most houses are small/medium/large in terms of area.
plt.figure()
sns.histplot(df["area"], kde=True)
plt.title("Distribution of House Area")
plt.xlabel("Area")
plt.ylabel("Count")
plt.show()
```

```python
# 3) Relationship between area and price
# Thought process: visually check whether larger houses tend to have higher prices.
plt.figure()
sns.scatterplot(data=df, x="area", y="price")
plt.title("House Price vs Area")
plt.xlabel("Area")
plt.ylabel("Price")
plt.show()
```

```python
# 4) Boxplots: how price varies with number of bedrooms
# Thought process: compare typical price ranges for different bedroom counts.
plt.figure()
sns.boxplot(data=df, x="bedrooms", y="price")
plt.title("Price by Number of Bedrooms")
plt.xlabel("Bedrooms")
plt.ylabel("Price")
plt.show()
```

```python
# 5) Boxplots: how price differs by furnishing status
# Thought process: see if furnished houses systematically cost more.
plt.figure()
sns.boxplot(data=df, x="furnishingstatus", y="price")
plt.title("Price by Furnishing Status")
plt.xlabel("Furnishing Status")
plt.ylabel("Price")
plt.xticks(rotation=15)
plt.show()
```

```python
# 6) Boxplots: price vs access to main road
# Thought process: test whether houses near the main road have different prices. [web:18][web:16]
plt.figure()
sns.boxplot(data=df, x="mainroad", y="price")
plt.title("Price by Main Road Access")
plt.xlabel("Mainroad (yes/no)")
plt.ylabel("Price")
plt.show()
```

```python
# 7) Correlation heatmap for numeric variables
# Thought process: identify which numeric features are most strongly related to price
# and to each other, which will be helpful for later modeling.
plt.figure(figsize=(8, 6))
corr = df[numeric_cols].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", center=0)
plt.title("Correlation Heatmap (Numeric Features)")
plt.show()
```

With your EDA complete, answer the following questions.

1. Was there anything surprising about your dataset? 
This housing prices dataset shows several interesting patterns that emerged during exploratory data analysis. The distribution of both price and area is strongly right‑skewed, with most houses clustered at relatively modest sizes and prices, and a small number of very large, very expensive properties standing out as potential outliers. Categorical features such as hot water heating and guestroom appear much less common than others, which makes those categories visually distinctive in frequency tables and plots and suggests that they may play a smaller but still notable role in price differences. Furnishing status also seems important: furnished houses generally show higher typical prices than semi‑furnished or unfurnished ones, hinting that furnishing could be a meaningful driver of value.


2. Do you have any concerns about your dataset? 
There are a few concerns to keep in mind when working with this dataset. The overall size is relatively small, which limits how complex or generalizable any predictive model can be and raises the risk of overfitting if too many features are included. Some categories in the yes/no variables are quite rare, meaning that the model may not learn stable patterns for those groups and any conclusions about them should be interpreted cautiously. In addition, several structural variables, such as area, number of bedrooms, and number of stories, appear related to each other, which could introduce multicollinearity issues for certain algorithms and may require careful feature selection or regularization.


3. Is there anything you want to make note of for the next phase of your analysis, which is cleaning data?
For the data‑cleaning phase, it will be important to address these issues systematically. Skewed numeric features like price and area may benefit from transformations (for example, taking the logarithm) to make relationships more linear and reduce the influence of extreme values.
